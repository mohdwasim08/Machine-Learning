## Column Transformer 
* ColumnTransformer applies different preprocessing methods to different columns.
* It is used when a dataset contains mixed data types (numerical and categorical).

```python
from sklearn.compose import ColumnTransformer 
```

```python
transformer = ColumnTransformer(transformers=[
    ('trf1', SimpleImputer(),['fever']),
    ('trf2', OrdinalEncoder(categories=[['Mild','Strong']]),['cough']),
    ('trf3', OneHotEncoder(sparse_output=False,drop='first'),['gender','city'])
],remainder='passthrough') # remainder--> keeps columns that are not transformed 
```

```python
trf = transformer.fit_transform(X_train)
```
